### B1. Propagación de la varianza
Construir una red de 50 capas de ancho 256 con distintas inicializaciones ($s=0.01$, $s^2=1/n$, $s^2=2/n$, $s=1$) y activaciones (tanh, ReLU). Graficar la desviación estándar de las activaciones por capa.

```python
def propaga(act,s2,L=50,n=256,N=1000):
    A=rng.standard_normal((N,n)); stds=[]
    for _ in range(L):
        W=rng.standard_normal((n,n))*np.sqrt(s2(n))
        A=act(A@W); stds.append(A.std())
    return stds
relu=lambda t:np.maximum(t,0)
inits={'s=0.01':lambda n:1e-4,'LeCun 1/n':lambda n:1/n,'He 2/n':lambda n:2/n,'s=1':lambda n:1.0}
fig,ax=plt.subplots(1,2,figsize=(11,3.5))
for k,(nombre,act) in enumerate([('tanh',np.tanh),('relu',relu)]):
    for iname,s2 in inits.items(): ax[k].semilogy(np.maximum(propaga(act,s2),1e-30),label=iname)
    ax[k].set_title(nombre); ax[k].legend()
plt.show()
```
