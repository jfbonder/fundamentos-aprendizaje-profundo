### B1. Interpolación lineal a trozos con ReLU (A4)
Construir explícitamente $u_N(x)=f(0)+\sum_{k=0}^{N-1}c_k\,\mathrm{ReLU}(x-k/N)$ con $c_k$ las diferencias de pendientes. Graficar el error en norma infinito en función de $N$ (log-log) para $f=\sin(2\pi x)$ y $f=\sqrt x$ y ajustar la tasa.

```python
relu=lambda t:np.maximum(t,0)
def interp_relu(f,N,x):
    nodos=np.linspace(0,1,N+1); pend=np.diff(f(nodos))*N
    c=np.r_[pend[0],np.diff(pend)]
    return f(0.0)+sum(ck*relu(x-nk) for ck,nk in zip(c,nodos[:-1]))
x=np.linspace(0,1,5001)
Ns=[2,4,8,16,32,64,128]
for nombre,f in [('sin(2 pi x)',lambda t:np.sin(2*np.pi*t)),('sqrt(x)',np.sqrt)]:
    errs=[np.abs(f(x)-interp_relu(f,N,x)).max() for N in Ns]
    plt.loglog(Ns,errs,'o-',label=nombre)
    print(nombre,'pendiente aprox:',np.polyfit(np.log(Ns),np.log(errs),1)[0])
plt.legend(); plt.show()
```
