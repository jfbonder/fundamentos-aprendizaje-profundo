### B1. Neurona sigmoidea con costo cuadrático vs. entropía cruzada
Implementar el entrenamiento de una neurona en 1D con ambos costos, partiendo de $w_0=-5,b_0=0$ para el dato $x=1,y=1$. Graficar el costo por iteración para ambos.

```python
sigma=lambda t:1/(1+np.exp(-t))
def entrenar(costo,eta,n=300,w=-5.0,b=0.0,x=1.0,y=1.0):
    h=[]
    for _ in range(n):
        z=sigma(w*x+b)
        if costo=='mse':
            d=(z-y)*z*(1-z); h.append(0.5*(z-y)**2)
        else:
            d=(z-y); h.append(-np.log(z if y==1 else 1-z))
        w-=eta*d*x; b-=eta*d
    return h
for c in ['mse','ce']:
    plt.plot(entrenar(c,0.5),label=c)
plt.legend(); plt.yscale('log'); plt.show()
```
