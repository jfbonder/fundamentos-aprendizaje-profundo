```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=(6,4)
rng=np.random.default_rng(0)
```

### B1. Cuadrados mínimos
Generar $y=2x-1+\varepsilon$, $\varepsilon\sim N(0,0.3^2)$. Resolver las ecuaciones normales con `np.linalg.solve` y comparar con `np.polyfit` y `np.linalg.lstsq`.

```python
N=50
x=rng.uniform(-1,1,N); y=2*x-1+0.3*rng.standard_normal(N)
A=np.column_stack([x,np.ones(N)])
w,b=np.linalg.solve(A.T@A,A.T@y)
print(w,b, np.polyfit(x,y,1))
plt.scatter(x,y); xs=np.linspace(-1,1,2); plt.plot(xs,w*xs+b,'r'); plt.show()
```

### B2. Paisaje de la función de costo
Graficar curvas de nivel de $C(w,b)$ y superponer el gradiente (`plt.quiver`). Verificar que el gradiente es ortogonal a las curvas de nivel.

```python
W,B=np.meshgrid(np.linspace(0,4,100),np.linspace(-3,1,100))
Cost=0.5*((W[...,None]*x+B[...,None]-y)**2).sum(-1)
plt.contour(W,B,Cost,30)
gw=((W[...,None]*x+B[...,None]-y)*x).sum(-1); gb=((W[...,None]*x+B[...,None]-y)).sum(-1)
s=slice(None,None,10)
plt.quiver(W[s,s],B[s,s],gw[s,s],gb[s,s],color='r'); plt.xlabel('w'); plt.ylabel('b'); plt.show()
```
